```python
#loading data
```

```python
import pandas as pds
df = pd.read_csv('/content/movies.csv')
```

```python
df.head()
```

```python
df.columns
```

```python
#preprocessing
```

```python
df.drop(columns=['index', 'homepage', 'id', 'keywords',"budget",
       'original_language', 'original_title', 'overview', 'popularity',
       'production_companies', 'production_countries', 'release_date',
       'revenue', 'runtime', 'status', 'tagline'
       , 'vote_count', 'cast'],axis=1,inplace=True)
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.head()
```

```python
df.rename(columns={'vote_average': 'rating'}, inplace=True)
```

```python
df.head()
```

```python
#making data ready for transformation
```

```python
df['features']=df['genres']+' '+df['director']+' '+df['spoken_languages']+' '+df['rating'].astype(str)
```

```python
df['features'].head()
```

```python
#converting data to numerical format so that model can understand
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vect=TfidfVectorizer(stop_words='english')
features_matrix=vect.fit_transform(df['features'])
```

```python
print(features_matrix.shape)
```

```python
#making similarity matrix
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity_matrix=cosine_similarity(features_matrix)
```

```python
#function for recommendation
```

```python
def recommend(movie_name, top_n=5):
    if movie_name not in df["title"].values:
        return "Movie not found"

    movie_index = df[df["title"] == movie_name].index[0]

    similarity_scores = list(enumerate(similarity_matrix[movie_index]))
    similarity_scores = sorted(similarity_scores, key=lambda x: x[1], reverse=True)

    recommendations = []
    for i in similarity_scores[1:top_n+1]:
        recommendations.append(df.iloc[i[0]]["title"])

    return recommendations
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity_matrix=cosine_similarity(features_matrix)
```
